## 문제 1. 은닉층 1개 모델 (25점)
`nn.Module` 을 상속해 아래 조건의 모델을 작성하세요.

- 입력 특성 5개, 은닉층 뉴런 16개, 출력 3개
- `__init__` 과 `forward` 를 모두 구현하고 각 줄에 주석
- `print(model)` 로 구조 출력


```python
# 답안
import torch
import torch.nn as nn

# nn.Module을 상속받아 신경망 클래스 정의
class SingleHiddenLayerNet(nn.Module):
    def __init__(self):
        # 부모 클래스(nn.Module)의 생성자 호출하여 초기화
        super(SingleHiddenLayerNet, self).__init__()

        # 입력 특성 5개 -> 은닉층 뉴런 16개로 연결하는 선형 레이어 정의
        self.fc1 = nn.Linear(in_features=5, out_features=16)

        # 은닉층 활성화 함수로 ReLU 정의
        self.relu = nn.ReLU()

        # 은닉층 뉴런 16개 -> 출력 뉴런 3개로 연결하는 선형 레이어 정의
        self.fc2 = nn.Linear(in_features=16, out_features=3)

    def forward(self, x):
        # 입력 데이터 x를 첫 번째 선형 레이어(fc1)에 전달
        out = self.fc1(x)

        # 선형 레이어의 출력을 활성화 함수(ReLU)에 전달
        out = self.relu(out)

        # 활성화 함수의 출력을 두 번째 선형 레이어(fc2)에 전달하여 최종 출력 계산
        out = self.fc2(out)

        # 최종 계산된 출력 값 반환
        return out

# 모델 객체 생성
model = SingleHiddenLayerNet()

# model 구조 출력
print(model)
```

## 문제 2. 은닉층 2개 모델 (25점)
문제 1과 같은 입력·출력에 은닉층을 2개로 늘린 모델을 작성하고 구조를 출력하세요.


```python
# 답안
import torch
import torch.nn as nn

# nn.Module을 상속받아 은닉층 2개의 신경망 클래스 정의
class TwoHiddenLayerNet(nn.Module):
    def __init__(self):
        # 부모 클래스(nn.Module)의 생성자 호출하여 초기화
        super(TwoHiddenLayerNet, self).__init__()

        # 입력 특성 5개 -> 첫 번째 은닉층 뉴런 16개로 연결하는 선형 레이어 정의
        self.fc1 = nn.Linear(in_features=5, out_features=16)

        # 첫 번째 활성화 함수로 ReLU 정의
        self.relu1 = nn.ReLU()

        # 첫 번째 은닉층 16개 -> 두 번째 은닉층 뉴런 8개로 연결하는 선형 레이어 정의
        self.fc2 = nn.Linear(in_features=16, out_features=8)

        # 두 번째 활성화 함수로 ReLU 정의
        self.relu2 = nn.ReLU()

        # 두 번째 은닉층 8개 -> 최종 출력 뉴런 3개로 연결하는 선형 레이어 정의
        self.fc3 = nn.Linear(in_features=8, out_features=3)

    def forward(self, x):
        # 입력 데이터 x를 첫 번째 선형 레이어(fc1)에 전달
        out = self.fc1(x)

        # 첫 번째 활성화 함수(ReLU) 적용
        out = self.relu1(out)

        # 두 번째 선형 레이어(fc2)에 전달
        out = self.fc2(out)

        # 두 번째 활성화 함수(ReLU) 적용
        out = self.relu2(out)

        # 세 번째 선형 레이어(fc3)에 전달하여 최종 출력 계산
        out = self.fc3(out)

        # 최종 계산된 출력 값 반환
        return out

# 모델 객체 생성
model = TwoHiddenLayerNet()

# model 구조 출력
print(model)
```

## 문제 5. 임의 입력으로 확인 (보너스)
두 모델에 각각 `torch.randn(8, 5)` 를 넣어 출력 shape을 확인하고, 중간 층의 shape 변화를 출력하세요.


```python
# 답안
import torch
import torch.nn as nn

# 1. 은닉층 1개 모델 정의
class SingleHiddenLayerNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(5, 16)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(16, 3)

    def forward(self, x):
        print(f"[은닉층 1개 모델] 입력 Shape: {x.shape}")
        out = self.fc1(x)
        print(f"  -> fc1 통과 후: {out.shape}")
        out = self.relu(out)
        out = self.fc2(out)
        print(f"  -> 최종 출력 Shape: {out.shape}\n")
        return out

# 2. 은닉층 2개 모델 정의
class TwoHiddenLayerNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(5, 16)
        self.relu1 = nn.ReLU()
        self.fc2 = nn.Linear(16, 8)
        self.relu2 = nn.ReLU()
        self.fc3 = nn.Linear(8, 3)

    def forward(self, x):
        print(f"[은닉층 2개 모델] 입력 Shape: {x.shape}")
        out = self.fc1(x)
        print(f"  -> fc1 통과 후: {out.shape}")
        out = self.relu1(out)
        out = self.fc2(out)
        print(f"  -> fc2 통과 후: {out.shape}")
        out = self.relu2(out)
        out = self.fc3(out)
        print(f"  -> 최종 출력 Shape: {out.shape}\n")
        return out

# 임의의 입력 데이터 생성 (배치 크기 8, 특성 5개)
x = torch.randn(8, 5)

# 모델 생성 및 포워드 연산 진행
model1 = SingleHiddenLayerNet()
model2 = TwoHiddenLayerNet()

model1(x)
model2(x)
```
